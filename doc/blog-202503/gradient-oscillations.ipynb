## Visualisation Tools

Before getting started, setup the visualisations tools.

### Gradient correlations
One way to detect oscillations is to track the Pearson correlation of the gradients from epoch-to-epoch.

In my own experiments I've found that doing this epoch-to-epoch provides much cleaner, more interpretable results than doing the same step-to-step. The problem with step-to-step correlations is that there is a high variance due to batch-to-batch variations, as is the fundamental idea behind stochastic gradient descent. But we want to measure the lower-frequency oscillations that span multiple steps. For a training procedure with few epochs, the sweetspot is somewhere between per-epoch and per-step correlations, something that the approach here cannot achieve.

Inspiration:
* Morchdi, C., Zhou, Y., Ding, J., & Wang, B. (2023). Exploring Gradient Oscillation in Deep Neural Network Training. 59th Annual Allerton Conference on Communication, Control, and Computing (Allerton, USA). https://www.sci.utah.edu/~beiwang/publications/Mysterious_BeiWang_2023.pdf


```python
def plot_correlation_trends(gradients_cb, title="Correlation Between Consecutive Training Epochs", xlabel="epoch"):
  """Works against per-epoch and per-step data, although in practice the results are pretty meaningless when plotting against per-step data"""

  # validate
  if gradients_cb.gradients is None:
    raise ValueError("Cannot plot - raw gradients were not collected")

  # prepare
  # - identify which variables we're plotting for
  variable_index_by_layer = tinstr.variable_indices_by_layer(gradients_cb.model, single=True)
  variable_indices = [var_idx for var_idx in variable_index_by_layer if var_idx is not None]
  layer_indices = [l_idx for l_idx, var_idx in enumerate(variable_index_by_layer) if var_idx is not None]
  chart_names = [f"layer {l_idx}:\n{model.variables[var_idx].path}" for var_idx, l_idx in zip(variable_indices, layer_indices)]
  num_layers = len(layer_indices)

  # start figure
  # - at most 4 wide
  grid_width = 4
  grid_height = int(np.ceil(num_layers / grid_width))
  plt.figure(figsize=(15, grid_height * 3), layout='constrained')

  # chart per variable
  for i, var_idx in enumerate(variable_indices):
    # compute correlation between consecutive iterations
    data = np.stack(gradients_cb.gradients[var_idx], axis=0)  # shape: (iterations, spatial-dims...)
    data = data.reshape(data.shape[0], -1)  # shape: (iterations, spatial-size), where spatial-size = scalar flattened size
    iterations = data.shape[0]  # Number of epochs (or steps)
    correlations = [scipy.stats.pearsonr(data[i-1], data[i])[0] for i in range(1, iterations)]
    correlations = np.array(correlations)

    # Do chart
    plt.subplot(grid_height, grid_width, i + 1)
    plt.title(chart_names[i])
    plt.plot(range(1, iterations), correlations)
    plt.xlabel(xlabel)
    plt.ylabel("correlation (ep-1 → ep)")
    plt.margins(0)
    plt.ylim([-1.1, 1.1])
    plt.gca().xaxis.set_major_locator(mticker.MaxNLocator(integer=True))  # ensure integer x-axis ticks
  plt.show()
```

### Weight PCA
Another way to visualise the learning trajectory is to run PCA over the weights at each step or epoch. This can sometimes highlight whether oscillations are a problem or not.

The following implementation also enables the ability to experiment a bit with the PCA by selecting arbitrary axes and by choosing whether to normalize the data. When normalizing, we don't change the means as we still want a zero weight to reflect as a zero value. In general, I think that normalizing weights for this purpose is (a) superfluous, because the weights are already in a fairly normalized form through their initialization schemes, and (b) in appropriate, because it discards important information about the relative scales of different weight entries.

A related approach is to fit a smoothed curve to the raw data, and then compare the raw data against the smoothed curve. This gives a delta under the same dimensionality. By plotting the first principle component of the delta we get a signed magnitude of deviation, suitable for plotting. This is similar to plotting the norm of the delta, but the sign means that we can still identify oscillations where the magnitude of the norm vector remains the same but the angle changes.

This latter approach is closely related to the gradient correlations approach above, except that it makes it possible to measure the effect of oscillations near the per-step frequency. In the code below I default to smoothing sigma=steps-per-epoch, which I've found experimentally to provide a good balance when plotting on per-step data. However there is a disadvantage with this approach in the amount of data it holds in memory, and the computational complexity of PCA. A better overall solution likely combines the smoothing trick with the use of the correlation measure. The smoothing can be done via an exponential running average, eliminating the need to hold the full trajectory of weight data.

Inspiration:

* Visualizing the Loss Landscape of Neural Nets. NIPS 2018.
https://arxiv.org/abs/1712.09913

```python
def plot_weight_trajectories(variables_cb, axes=None, normalize=False):
  """
  Works against per-epoch and per-step data, but it's best viewed against per-step data.
  Falls back to a simple linear plot if there's only one axes (eg: for binary classification output layers).
  Args:
    axes - list of 2-axes to plot. Default: [0, 1]
  """
  if axes is None:
    axes = [0, 1]
  if len(axes) != 2:
    raise ValueError("axes must be a list of 2 axes")

  # validate
  if variables_cb.variables is None:
    raise ValueError("Cannot plot - raw variables were not collected")

  # prepare
  # - identify the variables which have data
  # - then identify the main variables of each layer
  # - take the logical interset as the variables that we'll plot.
  available_variable_indices = [v_idx for v_idx, v_data in enumerate(variables_cb.variables) if v_data is not None]

  variable_index_by_layer = tinstr.variable_indices_by_layer(variables_cb.model, single=True)
  variable_indices = [var_idx for var_idx in variable_index_by_layer if var_idx is not None]
  layer_indices = [l_idx for l_idx, var_idx in enumerate(variable_index_by_layer) if var_idx is not None]
  filter = np.isin(variable_indices, available_variable_indices)
  variable_indices = np.array(variable_indices)[filter]
  layer_indices = np.array(layer_indices)[filter]
  chart_names = [f"layer {l_idx}:\n{model.variables[var_idx].path}" for var_idx, l_idx in zip(variable_indices, layer_indices)]
  num_layers = len(layer_indices)
  tlabel = 'step' if hasattr(variables_cb, 'steps') else 'epoch'

  # start figure
  # - at most 4 wide
  grid_width = 4
  grid_height = int(np.ceil(num_layers / grid_width))
  plt.figure(figsize=(15, grid_height * 3), layout='constrained')

  # chart per variable
  for i, var_idx in enumerate(variable_indices):
    # prepare data
    data = np.stack(variables_cb.variables[var_idx], axis=0)
    iterations = data.shape[0]
    data = data.reshape(data.shape[0], -1)

    if data.shape[1] >= 2:
      # limit to available axes
      if data.shape[1] < max(axes)+1:
        chart_axes = [data.shape[1]-2, data.shape[1]-1]
      else:
        chart_axes = axes

      # compute PCA of variable over time
      pca = sklearn.decomposition.PCA(n_components=max(chart_axes)+1)
      if normalize:
        scaler = sklearn.preprocessing.StandardScaler(with_mean=False).fit(data)
        transformed_data = scaler.transform(data)
      else:
        transformed_data = data
      transformed_data = pca.fit_transform(transformed_data)  # (step, n_components)

      # Do chart
      plt.subplot(grid_height, grid_width, i + 1)
      plt.title(chart_names[i])
      plt.scatter(transformed_data[:, chart_axes[0]], transformed_data[:, chart_axes[1]], c=range(iterations), cmap='viridis', alpha=0.1, marker='.')
      cbar = plt.colorbar(label=f"training {tlabel}")
      cbar.solids.set(alpha=1)
      plt.xlabel(f"weights (PC{chart_axes[0]+1})")
      plt.ylabel(f"weights (PC{chart_axes[1]+1})")
    else:
      # do simple linear plot for linear data
      plt.subplot(grid_height, grid_width, i + 1)
      plt.title(chart_names[i])
      plt.scatter(range(iterations), data[:, 0], c=range(iterations), cmap='viridis', alpha=0.1, marker='.')
      cbar = plt.colorbar(label=f"training {tlabel}")
      cbar.solids.set(alpha=1)
      plt.xlabel(f"training {tlabel}")
      plt.ylabel(f"weights (scalar)")
      plt.gca().xaxis.set_major_locator(mticker.MaxNLocator(integer=True))  # ensure integer x-axis ticks
  plt.show()
```

```python
def plot_relative_weight_trajectories(variables_cb, axes=None, normalize=False, reference_smoothness=1.0, ylim=None):
  """
  Works against per-epoch and per-step data, but it's best viewed against per-step data.
  Falls back to a simple linear plot if there's only one axes (eg: for binary classification output layers).
  Args:
    axes - list of 2-axes to plot. Default: [0, 1]
    reference_smoothness - determines the smoothing sigma. If plotting per-step, then used as a factor of the number of steps per epoch,
    otherwise taken as is.
  """
  if axes is None:
    axes = [0, 1]
  if len(axes) != 2:
    raise ValueError("axes must be a list of 2 axes")

  # validate
  if variables_cb.variables is None:
    raise ValueError("Cannot plot - raw variables were not collected")

  # prepare
  # - identify the variables which have data
  # - then identify the main variables of each layer
  # - take the logical interset as the variables that we'll plot.
  available_variable_indices = [v_idx for v_idx, v_data in enumerate(variables_cb.variables) if v_data is not None]

  variable_index_by_layer = tinstr.variable_indices_by_layer(variables_cb.model, single=True)
  variable_indices = [var_idx for var_idx in variable_index_by_layer if var_idx is not None]
  layer_indices = [l_idx for l_idx, var_idx in enumerate(variable_index_by_layer) if var_idx is not None]
  filter = np.isin(variable_indices, available_variable_indices)
  variable_indices = np.array(variable_indices)[filter]
  layer_indices = np.array(layer_indices)[filter]
  chart_names = [f"layer {l_idx}:\n{model.variables[var_idx].path}" for var_idx, l_idx in zip(variable_indices, layer_indices)]
  num_layers = len(layer_indices)
  tlabel = 'step' if hasattr(variables_cb, 'steps') else 'epoch'

  # start figure
  # - at most 4 wide
  grid_width = 4
  grid_height = int(np.ceil(num_layers / grid_width))
  plt.figure(figsize=(15, grid_height * 3), layout='constrained')

  # chart per variable
  for i, var_idx in enumerate(variable_indices):
    # prepare data
    data = np.stack(variables_cb.variables[var_idx], axis=0)
    iterations = data.shape[0]
    data = data.reshape(data.shape[0], -1)
    if hasattr(variables_cb, 'steps'):
      sigma = step_variables.params['steps'] * reference_smoothness
    else:
      sigma = reference_smoothness
    smoothed = scipy.ndimage.gaussian_filter1d(data, sigma=sigma, axis=0)
    data = data - smoothed

    if data.shape[1] >= 2:
      # compute PCA of variable over time
      pca = sklearn.decomposition.PCA(n_components=1)
      if normalize:
        scaler = sklearn.preprocessing.StandardScaler(with_mean=False).fit(data)
        transformed_data = scaler.transform(data)
      else:
        transformed_data = data
      transformed_data = pca.fit_transform(transformed_data)  # (step, n_components)

      # Do chart
      plt.subplot(grid_height, grid_width, i + 1)
      plt.title(chart_names[i])
      plt.xlabel(f"training {tlabel}")
      plt.ylabel(f"weight delta (PC1)")
      plt.plot(transformed_data[:, [0]])
      plt.ylim(ylim)
  plt.show()
```

## Oscillations due to Skewed Datasets

```python
# Load CIFAR-10 dataset
(x_train, y_train), _ = tf.keras.datasets.cifar10.load_data()
x_train = x_train.astype('float32') / 255.0

def cifar2_dataset(skewed=False):
  def filter(x, y):
    if y == 0:
      return True
    elif y == 1 and (not skewed or tf.random.uniform((1,))[0] < 0.1):
      # only include 10% of the time if skewed=True
      return True
    else:
      return False # skip all other classes

  tf.keras.utils.set_random_seed(1)
  dataset = tf.data.Dataset.from_tensor_slices((x_train, y_train))
  dataset = dataset.filter(filter).shuffle(buffer_size=1024, seed=1).batch(32)
  dataset = dataset.repeat().take(150) # avoid non-determinism in steps_per_epoch
  return dataset

def show_class_counts(dataset):
  cnt_0 = 0
  cnt_1 = 0
  cnt_b = 0
  for batch in dataset:
    y = batch[1]
    cnt_0 += tf.math.count_nonzero(y == 0).numpy()
    cnt_1 += tf.math.count_nonzero(y == 1).numpy()
    cnt_b += 1
  print(f"  {cnt_b} batches with Class 0: {cnt_0} entries, Class 1: {cnt_1} entries")

# Plot some examples
batch = next(iter(cifar2_dataset(skewed=True)))
plt.figure(figsize=(15,5))
for i in range(18):
    x, y = batch[0][i], batch[1][i]
    plt.subplot(3, 6, i + 1)
    plt.title(f"class {y}")
    plt.imshow(x)
    plt.axis('off')
plt.show()

print(f"Unskewed:")
show_class_counts(cifar2_dataset(skewed=False))
print(f"Skewed:")
show_class_counts(cifar2_dataset(skewed=True))
```

```python
def classification_model(classes=10):
    image_input = tf.keras.Input(shape=(32, 32, 3), name="image_input")

    # convolutional layers
    # - 32x32x3 -> 16x16x8 -> 8x8x16 -> 4x4x32 -> 2x2x64
    conv = layers.Conv2D(filters=8, kernel_size=3, activation='relu', padding='same', name='conv2d_1a')(image_input)
    conv = layers.Conv2D(filters=8, kernel_size=3, activation='relu', padding='same', name='conv2d_1b')(conv)
    conv = layers.MaxPooling2D(pool_size=2, name='pool2d_1')(conv)
    conv = layers.Conv2D(filters=16, kernel_size=3, activation='relu', padding='same', name='conv2d_2a')(conv)
    conv = layers.Conv2D(filters=16, kernel_size=3, activation='relu', padding='same', name='conv2d_2b')(conv)
    conv = layers.MaxPooling2D(pool_size=2, name='pool2d_2')(conv)
    conv = layers.Conv2D(filters=32, kernel_size=3, activation='relu', padding='same', name='conv2d_3a')(conv)
    conv = layers.Conv2D(filters=32, kernel_size=3, activation='relu', padding='same', name='conv2d_3b')(conv)
    conv = layers.MaxPooling2D(pool_size=2, name='pool2d_3')(conv)
    conv = layers.Conv2D(filters=64, kernel_size=3, activation='relu', padding='same', name='conv2d_4a')(conv)
    conv = layers.Conv2D(filters=64, kernel_size=3, activation='relu', padding='same', name='conv2d_4b')(conv)
    conv = layers.MaxPooling2D(pool_size=2, name='pool2d_4')(conv)

    # dense layers
    # - 2x2x64 -> 256 -> 128 -> 10
    out = layers.Flatten(name='x1_flatten')(conv)
    out = layers.Dense(128, activation='relu', name='x1_dense_1')(out)
    out = layers.Dense(128, activation='relu', name='head_dense1')(out)
    out = layers.Dense(128, activation='relu', name='head_dense2')(out)
    out = layers.Dense(classes, activation='linear', name='head_dense3')(out)  # logits output

    return tf.keras.Model(inputs=image_input, outputs=out)

# Uncomment to print model summary
#classification_model().summary()
```

**Experiment:**
* First, with unskewed data
